# Smart City Intelligence Advisory System

This notebook provides an interactive interface for getting weather reports, AQI data, smart city analytics, and health recommendations for any city using MCP (Model Context Protocol) servers.

```python
import nest_asyncio
nest_asyncio.apply()

import asyncio
import os
from fastmcp import Client
import ipywidgets as widgets
from IPython.display import display, clear_output
import threading
from datetime import datetime

print("All imports completed successfully!")
```

```python
class SmartCityAdvisor:
    """
    SmartCityAdvisor is responsible for interacting with Weather, AQI, LLM, and Smart City Intelligence MCP servers.
    """

    def __init__(self, weather_url: str = "http://localhost:8000", 
                 aqi_server_url: str = "http://localhost:8001", 
                 llm_server_url: str = "http://localhost:8002",
                 smart_city_url: str = "http://localhost:8003"):
        self.weather_url = weather_url
        self.aqi_server_url = aqi_server_url
        self.llm_server_url = llm_server_url
        self.smart_city_url = smart_city_url

    async def get_weather(self, location: str) -> str:
        try:
            async with Client(f"{self.weather_url}/sse") as client:
                return await client.call_tool("get_weather", {"location": location})
        except Exception as e:
            return f"Failed to get weather data for '{location}': {str(e)}"

    async def get_aqi_report(self, location: str) -> str:
        try:
            async with Client(f"{self.aqi_server_url}/sse") as client:
                result = await client.call_tool("get_aqi", {"location": location})
                return self._extract_text(result)
        except Exception as e:
            return f"Failed to get AQI report for '{location}': {str(e)}"

    async def get_health_recommendations(self, weather_report: str, aqi_report: str) -> str:
        try:
            async with Client(f"{self.llm_server_url}/sse") as client:
                result = await client.call_tool("safety_guidelines", {
                    "weather_report": weather_report,
                    "aqi_report": aqi_report
                })
                return self._extract_text(result)
        except Exception as e:
            return f"Failed to get safety recommendations: {str(e)}"

    async def get_traffic_analysis(self, location: str) -> str:
        try:
            async with Client(f"{self.smart_city_url}/sse") as client:
                result = await client.call_tool("get_traffic_congestion", {"location": location})
                return self._extract_text(result)
        except Exception as e:
            return f"Failed to get traffic analysis for '{location}': {str(e)}"

    async def get_energy_metrics(self, location: str) -> str:
        try:
            async with Client(f"{self.smart_city_url}/sse") as client:
                result = await client.call_tool("get_city_energy_metrics", {"location": location})
                return self._extract_text(result)
        except Exception as e:
            return f"Failed to get energy metrics for '{location}': {str(e)}"

    async def get_livability_index(self, location: str) -> str:
        try:
            async with Client(f"{self.smart_city_url}/sse") as client:
                result = await client.call_tool("get_urban_livability_index", {"location": location})
                return self._extract_text(result)
        except Exception as e:
            return f"Failed to get livability index for '{location}': {str(e)}"

    async def get_city_alerts(self, location: str) -> str:
        try:
            async with Client(f"{self.smart_city_url}/sse") as client:
                result = await client.call_tool("get_smart_city_alerts", {"location": location})
                return self._extract_text(result)
        except Exception as e:
            return f"Failed to get city alerts for '{location}': {str(e)}"

    def _extract_text(self, result) -> str:
        try:
            if isinstance(result, list):
                return "\n".join(block.text for block in result if hasattr(block, "text"))
            elif hasattr(result, "text"):
                return result.text
            return str(result)
        except Exception as e:
            return f"Failed to parse result: {str(e)}"

print("SmartCityAdvisor class defined successfully!")
print("Ready for comprehensive city intelligence analysis!")
```

```python
import re
import gradio as gr
import asyncio
import concurrent.futures
from datetime import datetime
import time

def detect_location_type(location_input: str) -> tuple[str, str]:
    """
    Detect if input is a zipcode/postal code or city name
    Returns: (location, input_type)
    """
    location = location_input.strip()
    
    # US zipcode patterns
    if re.match(r'^\d{5}(-\d{4})?$', location):
        return location, "US Zipcode"
    
    # UK postcode patterns
    if re.match(r'^[A-Z]{1,2}\d[A-Z\d]?\s*\d[A-Z]{2}$', location.upper()):
        return location.upper(), "UK Postcode"
    
    # Canadian postal code patterns
    if re.match(r'^[A-Z]\d[A-Z]\s*\d[A-Z]\d$', location.upper()):
        return location.upper(), "Canadian Postal Code"
    
    # Check for "zipcode, country" format
    if "," in location:
        parts = location.split(",")
        if len(parts) == 2:
            zip_part = parts[0].strip()
            country_part = parts[1].strip()
            
            # Check if first part looks like a zipcode
            if (re.match(r'^\d{5}(-\d{4})?$', zip_part) or  # US
                re.match(r'^[A-Z]{1,2}\d[A-Z\d]?\s*\d[A-Z]{2}$', zip_part.upper()) or  # UK
                re.match(r'^[A-Z]\d[A-Z]\s*\d[A-Z]\d$', zip_part.upper())):  # CA
                return location, f"International Format ({country_part})"
    
    # Default: treat as city name
    return location, "City Name"

def format_enhanced_report(weather_data, aqi_data, traffic_data, energy_data, livability_data, alerts_data, recommendations, location, location_type, timestamp):
    """Format comprehensive smart city intelligence report"""
    
    # Create location header with type detection
    location_info = f"**Location:** {location} ({location_type})"
    
    output = f"""# 🏙️ Smart City Intelligence Report

{location_info}  
**Generated:** {timestamp}  
**Analysis:** Comprehensive Urban Intelligence

---

## 🌤️ Weather Conditions

{weather_data}

---

## 🌬️ Air Quality Index

{aqi_data}

---

## 🚦 Traffic & Transportation Intelligence

{traffic_data}

---

## ⚡ Energy & Sustainability Metrics

{energy_data}

---

## 🏘️ Urban Livability Assessment

{livability_data}

---

## 🚨 Real-time City Alerts

{alerts_data}

---

## 🤖 AI Health & Safety Recommendations

{recommendations}

---

✅ **Analysis completed successfully at {datetime.now().strftime('%H:%M:%S')}**

🌍 **Location type: {location_type} • Powered by 4 specialized MCP servers**
"""
    return output

print("✅ Location detection and formatting functions defined!")
print("🌍 Now supports: US zipcodes, UK postcodes, Canadian postal codes, and international formats")
```

```python
import gradio as gr
import asyncio
import concurrent.futures
from datetime import datetime
import time
import re

def detect_location_type(location_input: str) -> tuple[str, str]:
    """Detect if input is a zipcode/postal code or city name"""
    location = location_input.strip()
    
    # US zipcode patterns
    if re.match(r'^\d{5}(-\d{4})?$', location):
        return location, "US Zipcode"
    
    # UK postcode patterns
    if re.match(r'^[A-Z]{1,2}\d[A-Z\d]?\s*\d[A-Z]{2}$', location.upper()):
        return location.upper(), "UK Postcode"
    
    # Canadian postal code patterns
    if re.match(r'^[A-Z]\d[A-Z]\s*\d[A-Z]\d$', location.upper()):
        return location.upper(), "Canadian Postal Code"
    
    # Check for "zipcode, country" format
    if "," in location:
        parts = location.split(",")
        if len(parts) == 2:
            zip_part = parts[0].strip()
            country_part = parts[1].strip()
            
            if (re.match(r'^\d{5}(-\d{4})?$', zip_part) or
                re.match(r'^[A-Z]{1,2}\d[A-Z\d]?\s*\d[A-Z]{2}$', zip_part.upper()) or
                re.match(r'^[A-Z]\d[A-Z]\s*\d[A-Z]\d$', zip_part.upper())):
                return location, f"International Format ({country_part})"
    
    return location, "City Name"

def process_smart_city_analysis(location_input, analysis_type):
    """Process smart city data with immediate results and LLM analysis"""
    
    if not location_input or not location_input.strip():
        return """# ⚠️ Input Required

Please enter a **city name** or **location**.

**Examples:**
- **Cities:** New York, London, Tokyo, Singapore
- **US Zipcodes:** 10001, 90210, 33101
- **UK Postcodes:** SW1A 1AA, EC1A 1BB, M1 1AA
- **Canadian Postal Codes:** M5V 3A8, K1A 0A6"""
    
    location, location_type = detect_location_type(location_input)
    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    start_time = time.time()
    
    def get_mcp_data():
        import asyncio
        
        async def fetch_mcp_data():
            agent = SmartCityAdvisor(
                "http://localhost:8000", 
                "http://localhost:8001", 
                "http://localhost:8002", 
                "http://localhost:8003"
            )
            
            try:
                if analysis_type == "Comprehensive Analysis":
                    # Get all MCP server data
                    weather_raw = await agent.get_weather(location)
                    weather_report = weather_raw[0].text if isinstance(weather_raw, list) else str(weather_raw)
                    
                    aqi_report = await agent.get_aqi_report(location)
                    traffic_analysis = await agent.get_traffic_analysis(location)
                    energy_metrics = await agent.get_energy_metrics(location)
                    livability_index = await agent.get_livability_index(location)
                    city_alerts = await agent.get_city_alerts(location)
                    
                    return {
                        'weather': weather_report,
                        'aqi': aqi_report,
                        'traffic': traffic_analysis,
                        'energy': energy_metrics,
                        'livability': livability_index,
                        'alerts': city_alerts
                    }
                    
                elif analysis_type == "Weather & Health":
                    weather_raw = await agent.get_weather(location)
                    weather_report = weather_raw[0].text if isinstance(weather_raw, list) else str(weather_raw)
                    aqi_report = await agent.get_aqi_report(location)
                    
                    return {
                        'weather': weather_report, 
                        'aqi': aqi_report
                    }
                    
                else:
                    # Single analysis modes
                    if analysis_type == "Traffic Analysis":
                        traffic_analysis = await agent.get_traffic_analysis(location)
                        return {'traffic': traffic_analysis}
                    elif analysis_type == "Energy Metrics":
                        energy_metrics = await agent.get_energy_metrics(location)
                        return {'energy': energy_metrics}
                    elif analysis_type == "Livability Index":
                        livability_index = await agent.get_livability_index(location)
                        return {'livability': livability_index}
                    elif analysis_type == "City Alerts":
                        city_alerts = await agent.get_city_alerts(location)
                        return {'alerts': city_alerts}
                        
            except Exception as e:
                return f"Error fetching data: {str(e)}"
        
        # Run async function
        try:
            loop = asyncio.get_event_loop()
            if loop.is_running():
                import threading
                result = None
                exception = None
                
                def run_in_thread():
                    nonlocal result, exception
                    try:
                        new_loop = asyncio.new_event_loop()
                        asyncio.set_event_loop(new_loop)
                        result = new_loop.run_until_complete(fetch_mcp_data())
                    except Exception as e:
                        exception = e
                    finally:
                        new_loop.close()
                
                thread = threading.Thread(target=run_in_thread)
                thread.start()
                thread.join(timeout=60)
                
                if exception:
                    raise exception
                return result
            else:
                return loop.run_until_complete(fetch_mcp_data())
        except Exception as e:
            return f"Async execution error: {str(e)}"
    
    def get_comprehensive_llm_analysis(mcp_results):
        """Get comprehensive LLM analysis with all smart city data"""
        import asyncio
        from fastmcp import Client
        
        async def fetch_comprehensive_analysis():
            try:
                async with Client("http://localhost:8002/sse") as client:
                    result = await client.call_tool("safety_guidelines", {
                        "weather_report": mcp_results.get('weather', ''),
                        "aqi_report": mcp_results.get('aqi', ''),
                        "traffic_report": mcp_results.get('traffic', ''),
                        "alerts_report": mcp_results.get('alerts', ''),
                        "livability_report": mcp_results.get('livability', ''),
                        "smart_city_report": mcp_results.get('energy', '')
                    })
                    
                    # Extract text from result
                    if isinstance(result, list):
                        return "\n".join(block.text for block in result if hasattr(block, "text"))
                    elif hasattr(result, "text"):
                        return result.text
                    return str(result)
                    
            except Exception as e:
                return f"Error getting AI analysis: {str(e)}"
        
        # Run async LLM function
        try:
            loop = asyncio.get_event_loop()
            if loop.is_running():
                import threading
                result = None
                exception = None
                
                def run_in_thread():
                    nonlocal result, exception
                    try:
                        new_loop = asyncio.new_event_loop()
                        asyncio.set_event_loop(new_loop)
                        result = new_loop.run_until_complete(fetch_comprehensive_analysis())
                    except Exception as e:
                        exception = e
                    finally:
                        new_loop.close()
                
                thread = threading.Thread(target=run_in_thread)
                thread.start()
                thread.join(timeout=90)
                
                if exception:
                    raise exception
                return result
            else:
                return loop.run_until_complete(fetch_comprehensive_analysis())
        except Exception as e:
            return f"AI analysis error: {str(e)}"
    
    try:
        # Get MCP server data
        with concurrent.futures.ThreadPoolExecutor() as executor:
            future = executor.submit(get_mcp_data)
            mcp_results = future.result(timeout=60)
        
        mcp_time = time.time() - start_time
        
        # Handle MCP errors
        if isinstance(mcp_results, str) and "Error" in mcp_results:
            return f"""# ❌ Data Collection Error

**Location:** {location} ({location_type})  
**Time:** {timestamp}

**Error Details:** {mcp_results}

**Please check MCP servers are running on ports 8000, 8001, 8002, 8003**
"""
        
        # Generate reports based on analysis type
        if analysis_type == "Comprehensive Analysis" and isinstance(mcp_results, dict):
            # Get AI analysis
            llm_start = time.time()
            
            with concurrent.futures.ThreadPoolExecutor() as executor:
                future = executor.submit(get_comprehensive_llm_analysis, mcp_results)
                ai_analysis = future.result(timeout=120)
            
            llm_time = time.time() - llm_start
            total_time = time.time() - start_time
            
            return f"""# 🏙️ Smart City Intelligence Report

**Location:** {location} ({location_type})  
**Generated:** {timestamp}  
**Processing Time:** {total_time:.1f} seconds

---

## 🌤️ Weather Conditions

{mcp_results['weather']}

---

## 🌬️ Air Quality Index

{mcp_results['aqi']}

---

## 🚦 Traffic & Transportation

{mcp_results['traffic']}

---

## ⚡ Energy & Sustainability

{mcp_results['energy']}

---

## 🏘️ Urban Livability

{mcp_results['livability']}

---

## 🚨 City Alerts

{mcp_results['alerts']}

---

## 🤖 AI Health & Safety Analysis

{ai_analysis}

---

✅ **Analysis completed successfully in {total_time:.1f} seconds**
"""
            
        elif analysis_type == "Weather & Health" and isinstance(mcp_results, dict):
            # Get AI analysis for weather & health
            llm_start = time.time()
            
            with concurrent.futures.ThreadPoolExecutor() as executor:
                future = executor.submit(get_comprehensive_llm_analysis, mcp_results)
                ai_analysis = future.result(timeout=90)
            
            llm_time = time.time() - llm_start
            total_time = time.time() - start_time
            
            return f"""# 🌍 Weather & Health Analysis

**Location:** {location} ({location_type})  
**Generated:** {timestamp}  
**Processing Time:** {total_time:.1f} seconds

---

## 🌤️ Weather Conditions

{mcp_results['weather']}

---

## 🌬️ Air Quality Index

{mcp_results['aqi']}

---

## 🤖 AI Health & Safety Recommendations

{ai_analysis}

---

✅ **Analysis completed successfully in {total_time:.1f} seconds**
"""
        
        else:
            # Single analysis modes
            data_key = list(mcp_results.keys())[0] if isinstance(mcp_results, dict) else 'data'
            data_value = list(mcp_results.values())[0] if isinstance(mcp_results, dict) else mcp_results
            
            return f"""# 🏙️ {analysis_type}

**Location:** {location} ({location_type})  
**Generated:** {timestamp}  
**Processing Time:** {mcp_time:.1f} seconds

---

{data_value}

---

✅ **{analysis_type} completed successfully**
"""
            
    except concurrent.futures.TimeoutError:
        return f"""# ⏰ Processing Timeout

**Location:** {location} ({location_type})  
**Time:** {timestamp}

The analysis is taking longer than expected. Please try again or check if all MCP servers are responsive.
"""
        
    except Exception as e:
        return f"""# ❌ System Error

**Location:** {location} ({location_type})  
**Time:** {timestamp}

**Error:** {str(e)}

**Please check all MCP servers are running on the required ports.**
"""

def create_smart_city_interface():
    """Create clean Smart City Intelligence interface"""
    
    with gr.Blocks(
        title="Smart City Intelligence System",
        theme=gr.themes.Soft(),
        css="""
        .gradio-container {
            font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
        }
        .main-header {
            background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
            color: white;
            padding: 30px;
            border-radius: 15px;
            text-align: center;
            margin-bottom: 25px;
        }
        """
    ) as demo:
        
        gr.HTML("""
        <div class="main-header">
            <h1 style="margin: 0; font-size: 2.5em;">🏙️ Smart City Intelligence System</h1>
            <p style="margin: 15px 0 5px 0; font-size: 1.2em;">Comprehensive Urban Analytics with AI-Powered Insights</p>
        </div>
        """)
        
        with gr.Row():
            with gr.Column(scale=3):
                location_input = gr.Textbox(
                    label="🌍 Location",
                    placeholder="Enter city name, zipcode, or postal code",
                    lines=1
                )
            
            with gr.Column(scale=2):
                analysis_type = gr.Dropdown(
                    label="🔍 Analysis Mode",
                    choices=[
                        "Comprehensive Analysis",
                        "Weather & Health", 
                        "Traffic Analysis",
                        "Energy Metrics",
                        "Livability Index",
                        "City Alerts"
                    ],
                    value="Comprehensive Analysis"
                )
            
            with gr.Column(scale=1):
                analyze_btn = gr.Button(
                    "🚀 Analyze",
                    variant="primary",
                    size="lg"
                )
        
        # Results display
        output_display = gr.Markdown(
            label="📋 Smart City Intelligence Report",
            value="""# 🏙️ Welcome to Smart City Intelligence

Enter a location above and select your analysis mode to get comprehensive urban analytics and AI-powered insights.

**Supported Locations:**
- **Cities:** London, New York, Tokyo, Singapore
- **US Zipcodes:** 10001, 90210, 33101
- **UK Postcodes:** SW1A 1AA, EC1A 1BB
- **Canadian Postal Codes:** M5V 3A8, K1A 0A6

**Analysis Modes:**
- **Comprehensive:** Complete city intelligence with AI recommendations
- **Weather & Health:** Environmental conditions with health insights
- **Traffic Analysis:** Real-time congestion and transportation
- **Energy Metrics:** Sustainability and energy data
- **Livability Index:** Urban quality of life assessment
- **City Alerts:** Real-time notifications and safety information

Ready to analyze your city! 🌟"""
        )
        
        # Event handlers
        analyze_btn.click(
            fn=process_smart_city_analysis,
            inputs=[location_input, analysis_type],
            outputs=[output_display]
        )
        
        location_input.submit(
            fn=process_smart_city_analysis,
            inputs=[location_input, analysis_type], 
            outputs=[output_display]
        )
        
        # Footer
        gr.HTML("""
        <div style="margin-top: 30px; padding: 20px; background: #f8f9fa; border-radius: 10px;">
            <div style="display: grid; grid-template-columns: repeat(auto-fit, minmax(300px, 1fr)); gap: 15px;">
                <div><strong>🌤️ Weather Intelligence:</strong> Real-time conditions and forecasts</div>
                <div><strong>🌬️ Air Quality Monitoring:</strong> Comprehensive AQI data and safety</div>
                <div><strong>🚦 Traffic Analytics:</strong> Congestion patterns and transportation</div>
                <div><strong>🤖 AI Analysis:</strong> Personalized health and safety recommendations</div>
            </div>
        </div>
        """)
    
    return demo

# Launch the application
try:
    demo = create_smart_city_interface()
    
    print("🚀 Launching Smart City Intelligence System...")
    print("📱 Application will be available at the links below:")
    
    demo.launch(
        share=True,
        server_name="0.0.0.0",
        server_port=7869,
        show_error=True,
        quiet=False
    )
    
except Exception as e:
    print(f"❌ Error launching system: {str(e)}")
    print("🔧 Please ensure all MCP servers are running on ports 8000, 8001, 8002, 8003")
```

```python
import gradio as gr
import asyncio
import concurrent.futures
from datetime import datetime
import time
import re

def detect_location_type(location_input: str) -> tuple[str, str]:
    """Detect if input is a zipcode/postal code or city name"""
    location = location_input.strip()
    
    # US zipcode patterns
    if re.match(r'^\d{5}(-\d{4})?$', location):
        return location, "US Zipcode"
    
    # UK postcode patterns
    if re.match(r'^[A-Z]{1,2}\d[A-Z\d]?\s*\d[A-Z]{2}$', location.upper()):
        return location.upper(), "UK Postcode"
    
    # Canadian postal code patterns
    if re.match(r'^[A-Z]\d[A-Z]\s*\d[A-Z]\d$', location.upper()):
        return location.upper(), "Canadian Postal Code"
    
    # Check for "zipcode, country" format
    if "," in location:
        parts = location.split(",")
        if len(parts) == 2:
            zip_part = parts[0].strip()
            country_part = parts[1].strip()
            
            if (re.match(r'^\d{5}(-\d{4})?$', zip_part) or
                re.match(r'^[A-Z]{1,2}\d[A-Z\d]?\s*\d[A-Z]{2}$', zip_part.upper()) or
                re.match(r'^[A-Z]\d[A-Z]\s*\d[A-Z]\d$', zip_part.upper())):
                return location, f"International Format ({country_part})"
    
    return location, "City Name"

def extract_detailed_infrastructure_info(livability_report: str) -> str:
    """Extract detailed facility information from livability report for LLM analysis"""
    
    # Extract facility details from the livability report
    detailed_info = "DETAILED INFRASTRUCTURE AND FACILITIES:\n\n"
    
    # Look for healthcare facilities section
    if "🏥 Healthcare Facilities" in livability_report:
        detailed_info += "HEALTHCARE FACILITIES:\n"
        lines = livability_report.split('\n')
        in_healthcare = False
        for line in lines:
            if "🏥 Healthcare Facilities" in line:
                in_healthcare = True
                continue
            elif in_healthcare and line.startswith('### ') and '🏥' not in line:
                in_healthcare = False
            elif in_healthcare and line.startswith('- **'):
                # Extract facility details
                detailed_info += line.replace('- **', '').replace('**', '') + "\n"
    
    # Look for educational institutions
    if "🏫 Educational Institutions" in livability_report:
        detailed_info += "\nEDUCATIONAL INSTITUTIONS:\n"
        lines = livability_report.split('\n')
        in_education = False
        for line in lines:
            if "🏫 Educational Institutions" in line:
                in_education = True
                continue
            elif in_education and line.startswith('### ') and '🏫' not in line:
                in_education = False
            elif in_education and line.startswith('- **'):
                detailed_info += line.replace('- **', '').replace('**', '') + "\n"
    
    # Look for shopping facilities
    if "🛒 Shopping & Retail" in livability_report:
        detailed_info += "\nSHOPPING & RETAIL FACILITIES:\n"
        lines = livability_report.split('\n')
        in_shopping = False
        for line in lines:
            if "🛒 Shopping & Retail" in line:
                in_shopping = True
                continue
            elif in_shopping and line.startswith('### ') and '🛒' not in line:
                in_shopping = False
            elif in_shopping and line.startswith('- **'):
                detailed_info += line.replace('- **', '').replace('**', '') + "\n"
    
    # Look for recreation facilities
    if "🏃 Recreation & Parks" in livability_report:
        detailed_info += "\nRECREATION & PARKS:\n"
        lines = livability_report.split('\n')
        in_recreation = False
        for line in lines:
            if "🏃 Recreation & Parks" in line:
                in_recreation = True
                continue
            elif in_recreation and line.startswith('### ') and '🏃' not in line:
                in_recreation = False
            elif in_recreation and line.startswith('- **'):
                detailed_info += line.replace('- **', '').replace('**', '') + "\n"
    
    # Look for road network details from traffic report
    return detailed_info

def extract_detailed_traffic_info(traffic_report: str) -> str:
    """Extract detailed road and transportation info from traffic report for LLM analysis"""
    
    detailed_info = "DETAILED TRANSPORTATION INFRASTRUCTURE:\n\n"
    
    # Extract major roads list
    if "Major Roads and Highways" in traffic_report:
        detailed_info += "MAJOR ROADS AND HIGHWAYS:\n"
        lines = traffic_report.split('\n')
        in_roads = False
        for line in lines:
            if "Major Roads and Highways" in line:
                in_roads = True
                continue
            elif in_roads and line.startswith('## '):
                in_roads = False
            elif in_roads and line.startswith('- **'):
                detailed_info += line.replace('- **', '').replace('**', '') + "\n"
    
    return detailed_info

def process_smart_city_analysis(location_input, analysis_type):
    """Process smart city data with detailed infrastructure information for LLM"""
    
    if not location_input or not location_input.strip():
        return """# ⚠️ Input Required

Please enter a **city name** or **location**.

**Examples:**
- **Cities:** New York, London, Tokyo, Singapore
- **US Zipcodes:** 10001, 90210, 33101
- **UK Postcodes:** SW1A 1AA, EC1A 1BB, M1 1AA
- **Canadian Postal Codes:** M5V 3A8, K1A 0A6"""
    
    location, location_type = detect_location_type(location_input)
    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    start_time = time.time()
    
    def get_mcp_data():
        import asyncio
        
        async def fetch_mcp_data():
            agent = SmartCityAdvisor(
                "http://localhost:8000", 
                "http://localhost:8001", 
                "http://localhost:8002", 
                "http://localhost:8003"
            )
            
            try:
                if analysis_type == "Comprehensive Analysis":
                    # Get all MCP server data
                    weather_raw = await agent.get_weather(location)
                    weather_report = weather_raw[0].text if isinstance(weather_raw, list) else str(weather_raw)
                    
                    aqi_report = await agent.get_aqi_report(location)
                    traffic_analysis = await agent.get_traffic_analysis(location)
                    energy_metrics = await agent.get_energy_metrics(location)
                    livability_index = await agent.get_livability_index(location)
                    city_alerts = await agent.get_city_alerts(location)
                    
                    return {
                        'weather': weather_report,
                        'aqi': aqi_report,
                        'traffic': traffic_analysis,
                        'energy': energy_metrics,
                        'livability': livability_index,
                        'alerts': city_alerts
                    }
                    
                elif analysis_type == "Weather & Health":
                    weather_raw = await agent.get_weather(location)
                    weather_report = weather_raw[0].text if isinstance(weather_raw, list) else str(weather_raw)
                    aqi_report = await agent.get_aqi_report(location)
                    
                    return {
                        'weather': weather_report, 
                        'aqi': aqi_report
                    }
                    
                else:
                    # Single analysis modes
                    if analysis_type == "Traffic Analysis":
                        traffic_analysis = await agent.get_traffic_analysis(location)
                        return {'traffic': traffic_analysis}
                    elif analysis_type == "Energy Metrics":
                        energy_metrics = await agent.get_energy_metrics(location)
                        return {'energy': energy_metrics}
                    elif analysis_type == "Livability Index":
                        livability_index = await agent.get_livability_index(location)
                        return {'livability': livability_index}
                    elif analysis_type == "City Alerts":
                        city_alerts = await agent.get_city_alerts(location)
                        return {'alerts': city_alerts}
                        
            except Exception as e:
                return f"Error fetching data: {str(e)}"
        
        # Run async function
        try:
            loop = asyncio.get_event_loop()
            if loop.is_running():
                import threading
                result = None
                exception = None
                
                def run_in_thread():
                    nonlocal result, exception
                    try:
                        new_loop = asyncio.new_event_loop()
                        asyncio.set_event_loop(new_loop)
                        result = new_loop.run_until_complete(fetch_mcp_data())
                    except Exception as e:
                        exception = e
                    finally:
                        new_loop.close()
                
                thread = threading.Thread(target=run_in_thread)
                thread.start()
                thread.join(timeout=60)
                
                if exception:
                    raise exception
                return result
            else:
                return loop.run_until_complete(fetch_mcp_data())
        except Exception as e:
            return f"Async execution error: {str(e)}"
    
    def get_comprehensive_llm_analysis(mcp_results):
        """Get comprehensive LLM analysis with detailed infrastructure information"""
        import asyncio
        from fastmcp import Client
        
        async def fetch_comprehensive_analysis():
            try:
                # Extract detailed infrastructure information
                detailed_infrastructure = ""
                detailed_transportation = ""
                
                if 'livability' in mcp_results:
                    detailed_infrastructure = extract_detailed_infrastructure_info(mcp_results['livability'])
                
                if 'traffic' in mcp_results:
                    detailed_transportation = extract_detailed_traffic_info(mcp_results['traffic'])
                
                # Combine detailed information with original reports
                enhanced_livability = mcp_results.get('livability', '') + "\n\n" + detailed_infrastructure
                enhanced_traffic = mcp_results.get('traffic', '') + "\n\n" + detailed_transportation
                
                async with Client("http://localhost:8002/sse") as client:
                    result = await client.call_tool("safety_guidelines", {
                        "weather_report": mcp_results.get('weather', ''),
                        "aqi_report": mcp_results.get('aqi', ''),
                        "traffic_report": enhanced_traffic,  # Enhanced with detailed road info
                        "alerts_report": mcp_results.get('alerts', ''),
                        "livability_report": enhanced_livability,  # Enhanced with detailed facility info
                        "smart_city_report": mcp_results.get('energy', '')
                    })
                    
                    # Extract text from result
                    if isinstance(result, list):
                        return "\n".join(block.text for block in result if hasattr(block, "text"))
                    elif hasattr(result, "text"):
                        return result.text
                    return str(result)
                    
            except Exception as e:
                return f"Error getting detailed AI analysis: {str(e)}"
        
        # Run async LLM function
        try:
            loop = asyncio.get_event_loop()
            if loop.is_running():
                import threading
                result = None
                exception = None
                
                def run_in_thread():
                    nonlocal result, exception
                    try:
                        new_loop = asyncio.new_event_loop()
                        asyncio.set_event_loop(new_loop)
                        result = new_loop.run_until_complete(fetch_comprehensive_analysis())
                    except Exception as e:
                        exception = e
                    finally:
                        new_loop.close()
                
                thread = threading.Thread(target=run_in_thread)
                thread.start()
                thread.join(timeout=90)
                
                if exception:
                    raise exception
                return result
            else:
                return loop.run_until_complete(fetch_comprehensive_analysis())
        except Exception as e:
            return f"Detailed AI analysis error: {str(e)}"
    
    try:
        # Get MCP server data
        with concurrent.futures.ThreadPoolExecutor() as executor:
            future = executor.submit(get_mcp_data)
            mcp_results = future.result(timeout=60)
        
        mcp_time = time.time() - start_time
        
        # Handle MCP errors
        if isinstance(mcp_results, str) and "Error" in mcp_results:
            return f"""# ❌ Data Collection Error

**Location:** {location} ({location_type})  
**Time:** {timestamp}

**Error Details:** {mcp_results}

**Please check MCP servers are running on ports 8000, 8001, 8002, 8003**
"""
        
        # Generate reports based on analysis type
        if analysis_type == "Comprehensive Analysis" and isinstance(mcp_results, dict):
            # Get AI analysis with detailed infrastructure info
            llm_start = time.time()
            
            with concurrent.futures.ThreadPoolExecutor() as executor:
                future = executor.submit(get_comprehensive_llm_analysis, mcp_results)
                ai_analysis = future.result(timeout=120)
            
            llm_time = time.time() - llm_start
            total_time = time.time() - start_time
            
            return f"""# 🏙️ Smart City Intelligence Report

**Location:** {location} ({location_type})  
**Generated:** {timestamp}  
**Processing Time:** {total_time:.1f} seconds

---

## 🌤️ Weather Conditions

{mcp_results['weather']}

---

## 🌬️ Air Quality Index

{mcp_results['aqi']}

---

## 🚦 Traffic & Transportation

{mcp_results['traffic']}

---

## ⚡ Energy & Sustainability

{mcp_results['energy']}

---

## 🏘️ Urban Livability

{mcp_results['livability']}

---

## 🚨 City Alerts

{mcp_results['alerts']}

---

## 🤖 AI Analysis with Detailed Infrastructure Insights

*This analysis considers specific healthcare facilities, schools, roads, shopping centers, and other infrastructure details.*

{ai_analysis}

---

✅ **Comprehensive analysis with detailed facility information completed in {total_time:.1f} seconds**
"""
            
        elif analysis_type == "Weather & Health" and isinstance(mcp_results, dict):
            # Get AI analysis for weather & health
            llm_start = time.time()
            
            with concurrent.futures.ThreadPoolExecutor() as executor:
                future = executor.submit(get_comprehensive_llm_analysis, mcp_results)
                ai_analysis = future.result(timeout=90)
            
            llm_time = time.time() - llm_start
            total_time = time.time() - start_time
            
            return f"""# 🌍 Weather & Health Analysis

**Location:** {location} ({location_type})  
**Generated:** {timestamp}  
**Processing Time:** {total_time:.1f} seconds

---

## 🌤️ Weather Conditions

{mcp_results['weather']}

---

## 🌬️ Air Quality Index

{mcp_results['aqi']}

---

## 🤖 AI Health & Safety Recommendations

{ai_analysis}

---

✅ **Analysis completed successfully in {total_time:.1f} seconds**
"""
        
        else:
            # Single analysis modes
            data_key = list(mcp_results.keys())[0] if isinstance(mcp_results, dict) else 'data'
            data_value = list(mcp_results.values())[0] if isinstance(mcp_results, dict) else mcp_results
            
            return f"""# 🏙️ {analysis_type}

**Location:** {location} ({location_type})  
**Generated:** {timestamp}  
**Processing Time:** {mcp_time:.1f} seconds

---

{data_value}

---

✅ **{analysis_type} completed successfully**
"""
            
    except concurrent.futures.TimeoutError:
        return f"""# ⏰ Processing Timeout

**Location:** {location} ({location_type})  
**Time:** {timestamp}

The detailed analysis is taking longer than expected. Please try again or check if all MCP servers are responsive.
"""
        
    except Exception as e:
        return f"""# ❌ System Error

**Location:** {location} ({location_type})  
**Time:** {timestamp}

**Error:** {str(e)}

**Please check all MCP servers are running on the required ports.**
"""

def create_smart_city_interface():
    """Create Smart City Intelligence interface with detailed infrastructure analysis"""
    
    with gr.Blocks(
        title="Smart City Intelligence System",
        theme=gr.themes.Soft(),
        css="""
        .gradio-container {
            font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
        }
        .main-header {
            background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
            color: white;
            padding: 30px;
            border-radius: 15px;
            text-align: center;
            margin-bottom: 25px;
        }
        """
    ) as demo:
        
        gr.HTML("""
        <div class="main-header">
            <h1 style="margin: 0; font-size: 2.5em;">🏙️ Smart City Intelligence System</h1>
            <p style="margin: 15px 0 5px 0; font-size: 1.2em;">Comprehensive Urban Analytics with Detailed Infrastructure Analysis</p>
        </div>
        """)
        
        with gr.Row():
            with gr.Column(scale=3):
                location_input = gr.Textbox(
                    label="🌍 Location",
                    placeholder="Enter city name, zipcode, or postal code",
                    lines=1
                )
            
            with gr.Column(scale=2):
                analysis_type = gr.Dropdown(
                    label="🔍 Analysis Mode",
                    choices=[
                        "Comprehensive Analysis",
                        "Weather & Health", 
                        "Traffic Analysis",
                        "Energy Metrics",
                        "Livability Index",
                        "City Alerts"
                    ],
                    value="Comprehensive Analysis"
                )
            
            with gr.Column(scale=1):
                analyze_btn = gr.Button(
                    "🚀 Analyze",
                    variant="primary",
                    size="lg"
                )
        
        # Results display
        output_display = gr.Markdown(
            label="📋 Smart City Intelligence Report",
            value="""# 🏙️ Welcome to Smart City Intelligence

Enter a location above and select your analysis mode to get comprehensive urban analytics with detailed infrastructure information.

** AI Analysis Includes:**
- **Specific Healthcare Facilities:** Hospital names, addresses, and types
- **Educational Institutions:** Schools, universities, and their locations  
- **Transportation Infrastructure:** Detailed road networks and public transport
- **Shopping & Services:** Retail centers, supermarkets, and service facilities
- **Recreation Facilities:** Parks, sports centers, and entertainment venues

**Supported Locations:**
- **Cities:** London, New York, Tokyo, Singapore
- **US Zipcodes:** 10001, 90210, 33101
- **UK Postcodes:** SW1A 1AA, EC1A 1BB
- **Canadian Postal Codes:** M5V 3A8, K1A 0A6

The AI now considers specific facility names and locations when providing health, safety, and lifestyle recommendations.

Ready to analyze your city with detailed infrastructure insights! 🌟"""
        )
        
        # Event handlers
        analyze_btn.click(
            fn=process_smart_city_analysis,
            inputs=[location_input, analysis_type],
            outputs=[output_display]
        )
        
        location_input.submit(
            fn=process_smart_city_analysis,
            inputs=[location_input, analysis_type], 
            outputs=[output_display]
        )
        
        # Footer
        gr.HTML("""
        <div style="margin-top: 30px; padding: 20px; background: #f8f9fa; border-radius: 10px;">
            <div style="display: grid; grid-template-columns: repeat(auto-fit, minmax(300px, 1fr)); gap: 15px;">
                <div><strong>🏥 Healthcare Analysis:</strong> Specific hospitals, clinics, and medical facilities</div>
                <div><strong>🏫 Education Insights:</strong> Schools, universities, and educational resources</div>
                <div><strong>🚦 Transportation Details:</strong> Named roads, highways, and public transport</div>
                <div><strong>🤖 Enhanced AI:</strong> Recommendations based on detailed infrastructure data</div>
            </div>
        </div>
        """)
    
    return demo

# Launch the application
try:
    demo = create_smart_city_interface()
    
    print("🚀 Launching Smart City Intelligence System with Detailed Infrastructure Analysis...")
    print("🏥 AI now analyzes specific healthcare facilities, schools, roads, and infrastructure details")
    print("📱 Application will be available at the links below:")
    
    demo.launch(
        share=True,
        server_name="0.0.0.0",
        server_port=7870,
        show_error=True,
        quiet=False
    )
    
except Exception as e:
    print(f"❌ Error launching system: {str(e)}")
    print("🔧 Please ensure all MCP servers are running on ports 8000, 8001, 8002, 8003")
```
